```python
import math
import numpy as np
```

```python
matrix = np.array([[1, 2,4],[2,3,11]])
print(matrix, "\n")

transposed = matrix.T
print(transposed, "\n")

# matrix multiplication

answer = matrix @ transposed

answer_ = transposed@matrix

print(answer, "\n \n", answer_ , '\n')
```

```python
a = np.array([[4,1], [4,7]])


b = np.array([[1,8],[0,2]]).transpose()

print(a , '\n \n ', b , '\n \n', a @ b)
```
