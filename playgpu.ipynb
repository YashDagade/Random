```python
import torch
import time

# Check if CUDA is available for RTX 2080Ti or use MPS for Mac
device = torch.device("cuda" if torch.cuda.is_available() else "mps")

# Set the parameters
input_size = 4096
hidden_size = 2048
output_size = 1024
batch_size = 64
num_batches = 100
num_epochs = 5

# Create a random dataset
data = torch.randn(batch_size, input_size).to(device)
target = torch.randn(batch_size, output_size).to(device)

# Define a deep neural network
class DeepNet(torch.nn.Module):
    def __init__(self):
        super(DeepNet, self).__init__()
        self.fc1 = torch.nn.Linear(input_size, hidden_size)
        self.fc2 = torch.nn.Linear(hidden_size, hidden_size)
        self.fc3 = torch.nn.Linear(hidden_size, hidden_size)
        self.fc4 = torch.nn.Linear(hidden_size, output_size)
        self.relu = torch.nn.ReLU()

    def forward(self, x):
        x = self.relu(self.fc1(x))
        x = self.relu(self.fc2(x))
        x = self.relu(self.fc3(x))
        x = self.fc4(x)
        return x

# Initialize the network, loss function, and optimizer
model = DeepNet().to(device)
criterion = torch.nn.MSELoss()
optimizer = torch.optim.Adam(model.parameters())

# Benchmarking loop
start_time = time.time()
for epoch in range(num_epochs):
    for _ in range(num_batches):
        # Forward pass
        optimizer.zero_grad()
        output = model(data)
        
        # Compute loss
        loss = criterion(output, target)
        
        # Backward pass and optimization
        loss.backward()
        optimizer.step()

end_time = time.time()
print(f"Time taken on {device}: {end_time - start_time} seconds")

# Print device and memory information
if device.type == 'cuda':
    device_name = torch.cuda.get_device_name(0)
    total_memory = torch.cuda.get_device_properties(0).total_memory
    allocated_memory = torch.cuda.memory_allocated(0)
    cached_memory = torch.cuda.memory_reserved(0)
    print(f"Device: {device_name}")
    print(f"Total Memory: {total_memory / 1e9:.2f} GB")
    print(f"Allocated Memory: {allocated_memory / 1e9:.2f} GB")
    print(f"Cached Memory: {cached_memory / 1e9:.2f} GB")
elif device.type == 'mps':
    print("Device: Apple MPS")
    # MPS doesn't support the same memory functions, so we'll just identify the device.
else:
    print("Running on CPU - no specific device information available.")
```
